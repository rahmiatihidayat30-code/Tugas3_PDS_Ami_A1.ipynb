### 4c. Dari data dictionary ke aturan validasi
Ubah isian `nilai_valid` Anda menjadi **minimal 3 aturan** yang bisa dijalankan:
- 1 aturan **rentang numerik**,
- 1 aturan **himpunan kategori**,
- 1 aturan **konsistensi antarkolom** (mis. kecepatan implisit untuk moda tertentu).

Gunakan kolom yang **berbeda** dari contoh demo dosen bila memungkinkan.

```python

# BAGIAN 4C: ATURAN VALIDASI

aturan_saya = {
    # Aturan rentang numerik
    "jarak_tidak_negatif":
        lambda d: d["jarak_km"].notna() & (d["jarak_km"] >= 0),

    # Aturan kategori
    "semester_sesuai_cakupan":
        lambda d: d["semester"].isin([3, 5]),

    # Aturan konsistensi antarkolom
    # Jika jarak = 0, waktu tempuh juga diharapkan = 0.
    "konsistensi_jarak_waktu":
        lambda d: (
            d["jarak_km"].notna()
            & d["waktu_tempuh_menit"].notna()
            & ~(
                (d["jarak_km"] == 0)
                & (d["waktu_tempuh_menit"] > 0)
            )
        )
}

for nama_aturan, fungsi in aturan_saya.items():
    lolos = fungsi(df).fillna(False)
    print(
        f"{nama_aturan}: "
        f"{int((~lolos).sum())} baris tidak lolos"
    )

    # Gunakan indeks baris, bukan nama responden,
    # untuk mengurangi paparan identitas.
    print("Indeks baris yang perlu diperiksa:",
          df.index[~lolos].tolist())
    print()
```

**Jawaban Anda:** Validasi dilakukan dengan memeriksa rentang jarak perjalanan, kesesuaian semester dengan cakupan survei, dan konsistensi antara jarak dengan waktu tempuh. Baris yang tidak lolos aturan belum tentu merupakan kesalahan pengisian karena dapat terjadi akibat definisi variabel yang belum jelas, perbedaan kondisi perjalanan, atau nilai yang sah tetapi tidak umum. Oleh karena itu, hasil validasi perlu ditinjau kembali dengan memeriksa definisi pada kuesioner dan meminta klarifikasi melalui prosedur yang berwenang jika diperlukan. Pemeriksaan sebaiknya menggunakan kode responden atau indeks baris, bukan menyebarkan nama atau informasi pribadi mahasiswa. Jumlah pelanggaran yang sebenarnya akan ditentukan berdasarkan hasil eksekusi kode pada dataset.

```python

# UJI KUASI-IDENTITAS

kolom_kuasi = [
    "semester",
    "usia",
    "perangkat_utama",
    "moda_transportasi"
]

kolom_tersedia = [
    k for k in kolom_kuasi if k in df.columns
]

if len(kolom_tersedia) >= 2:
    ukuran_grup = (
        df.groupby(kolom_tersedia, dropna=False)
        .size()
    )

    grup_tunggal = int((ukuran_grup == 1).sum())
    jumlah_grup = len(ukuran_grup)

    print("Kolom yang digunakan:", kolom_tersedia)
    print("Jumlah kombinasi karakteristik:", jumlah_grup)
    print("Kombinasi yang hanya muncul sekali:", grup_tunggal)
else:
    print("Kolom yang tersedia belum cukup untuk uji kuasi-identitas.")
```
